## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains page-level search and content-performance observations. I will use actual values from the starter CSV to check whether there is enough observable variation in search exposure and clicks to justify investigating this lane further.

```python
import pandas as pd

DATA_PATH = "../../data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

print(f"Number of pages: {len(df):,}")
print(f"Number of columns: {df.shape[1]:,}")

zero_click_pages = df[
    (df["impressions_90d"] >= 100) &
    (df["clicks_90d"] == 0)
]

print(
    f"Pages with >=100 impressions and 0 clicks: "
    f"{len(zero_click_pages):,}"
)

print(
    f"Percentage of pages: "
    f"{len(zero_click_pages) / len(df) * 100:.2f}%"
)
```

```python
high_visibility_zero_click = df[
    (df["impressions_90d"] >= 250) &
    (df["clicks_90d"] == 0)
]

print(
    f"Pages with >=250 impressions and 0 clicks: "
    f"{len(high_visibility_zero_click):,}"
)

print(
    f"Average impressions per page: "
    f"{df['impressions_90d'].mean():.2f}"
)
```

The results show that the dataset contains a measurable group of pages with meaningful search exposure but no recorded clicks. This makes search-performance and content opportunity worth investigating.

However, these observations do not prove that these pages require a content refresh. A page can have low or zero clicks for several reasons, including search intent, ranking position, demand, or measurement limitations. Therefore, I will use these numbers as evidence that the problem is worth investigating, not as proof that a particular action will improve a page.
